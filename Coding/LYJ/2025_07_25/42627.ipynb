# 디스크 컨트롤러

### 문제 설명

하드디스크는 한 번에 하나의 작업만 수행할 수 있습니다. 디스크 컨트롤러를 구현하는 방법은 여러 가지가 있습니다. 이 문제에서는 우선순위 디스크 컨트롤러라는 가상의 장치를 이용한다고 가정합니다. 우선순위 

디스크 컨트롤러는 다음과 같이 동작합니다.

어떤 작업 요청이 들어왔을 때 작업의 번호, 작업의 요청 시각, 작업의 소요 시간을 저장해 두는 대기 큐가 있습니다. 처음에 이 큐는 비어있습니다.

디스크 컨트롤러는 하드디스크가 작업을 하고 있지 않고 대기 큐가 비어있지 않다면 가장 우선순위가 높은 작업을 대기 큐에서 꺼내서 하드디스크에 그 작업을 시킵니다. 이때, 작업의 소요시간이 짧은 것, 작업의 요청 시각이 빠른 것, 작업의 번호가 작은 것 순으로 우선순위가 높습니다.
하드디스크는 작업을 한 번 시작하면 작업을 마칠 때까지 그 작업만 수행합니다.
하드디스크가 어떤 작업을 마치는 시점과 다른 작업 요청이 들어오는 시점이 겹친다면 하드디스크가 작업을 마치자마자 디스크 컨트롤러는 요청이 들어온 작업을 대기 큐에 저장한 뒤 우선순위가 높은 작업을 대기 큐에서 꺼내서 하드디스크에 그 작업을 시킵니다. 또, 하드디스크가 어떤 작업을 마치는 시점에 다른 작업이 들어오지 않더라도 그 작업을 마치자마자 또 다른 작업을 시작할 수 있습니다. 이 과정에서 걸리는 시간은 없다고 가정합니다.

```python
import heapq

def solution(jobs):
    sec = 0               # 현재 시점 (현재 시간)
    duration = 0          # 총 요청부터 종료까지 걸린 시간의 합

    # 요청 시간을 기준으로 entry_heap을 생성 (min-heap)
    entry_heap = [(job[0], job) for job in jobs]
    heapq.heapify(entry_heap)

    while entry_heap:
        candidates = []   # 현재 시점에서 실행 가능한 작업들

        # entry_heap에서 현재 시간(sec) 이전에 요청된 작업을 candidates에 넣음
        while entry_heap:
            candidate = heapq.heappop(entry_heap)

            if candidate[0] > sec:
                # 요청 시간이 미래라면 아직 수행 불가능 → 다시 넣고 중단
                heapq.heappush(entry_heap, candidate)
                break
            else:
                # (작업시간, 전체작업정보) 형식으로 저장 → 작업시간 기준으로 정렬됨
                heapq.heappush(candidates, (candidate[1][1], candidate[1]))

        if not candidates:
            # 수행 가능한 작업이 없다면 시간을 다음 작업의 요청 시간으로 이동
            sec = entry_heap[0][0]
            continue

        # 작업시간이 가장 짧은 작업을 꺼냄 (SJF 방식)
        candidate = heapq.heappop(candidates)

        # 현재 시간 업데이트: 작업 시간만큼 소요
        sec += candidate[0]

        # 총 소요 시간 누적: 현재 시간 - 요청 시간
        duration += sec - candidate[1][0]

        # 남은 candidates는 다시 entry_heap에 돌려줌
        for c in candidates:
            heapq.heappush(entry_heap, (c[1][0], c[1]))

    # 평균 소요 시간 반환
    return int(duration / len(jobs))
```

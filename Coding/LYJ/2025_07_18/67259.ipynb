```python
from collections import deque

def solution(board):
    n = len(board)
    INF = float('inf')

    # 상, 하, 좌, 우 (0,1,2,3)
    directions = [(-1,0), (1,0), (0,-1), (0,1)]

    # 3차원 비용 배열: [y][x][direction]
    cost = [[[INF]*4 for _ in range(n)] for _ in range(n)] # 상하좌우 각각의 방향에 대한 비용
    queue = deque()

    # 초기 방향은 우(3), 하(1) 로 시작 가능 (코너가 아니므로 100)
    for dir_idx in [1, 3]:
        cost[0][0][dir_idx] = 0
        queue.append((0, 0, dir_idx, 0))  # y, x, 방향, 누적 비용

    while queue:
        y, x, dir_idx, cur_cost = queue.popleft()

        for new_dir, (dy, dx) in enumerate(directions):
            ny, nx = y + dy, x + dx
            if 0 <= ny < n and 0 <= nx < n and board[ny][nx] == 0:
                # 같은 방향 → 직선, 다른 방향 → 코너
                if new_dir == dir_idx:
                    new_cost = cur_cost + 100
                else:
                    new_cost = cur_cost + 600

                if cost[ny][nx][new_dir] > new_cost:
                    cost[ny][nx][new_dir] = new_cost
                    queue.append((ny, nx, new_dir, new_cost))

    return min(cost[n-1][n-1])
```

# 이분탐색 - 입국심사 

### 문제 설명 요약
n명의 사람이 입국 심사를 받아야 함

입국심사관마다 한 사람을 심사하는 데 걸리는 시간이 다름 (배열 times)

모든 사람이 입국 심사를 받는데 최소 몇 분이 걸리는지 구해야 함

```python
def solution(n, times):
    # 최소 및 최대 시간 설정
    min_time = 1
    max_time = max(times) * n
    
    answer = 0
    
    while min_time <= max_time:
        # 시간 효율성을 위해 중간값으로 이분 탐색 실행
        mid_time = (min_time + max_time) // 2
        
        # 현재 시간에 처리할 수 있는 사람 수 계산
        total_people = sum(mid_time // time for time in times)
        
        if total_people >= n:
            answer = mid_time
            max_time = mid_time - 1  # 더 적은 시간으로 시도
        else:
            min_time = mid_time + 1
            
    return answer
```

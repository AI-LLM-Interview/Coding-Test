56. 구간 병합
중간

주제
프리미엄 잠금 아이콘
회사
intervals[i] = [starti, endi]인 구간 배열이 주어졌을 때, 겹치는 모든 구간을 병합하고, 입력에 포함된 모든 구간을 포함하는 겹치지 않는 구간의 배열을 반환합니다.

예제 1:

입력: intervals = [[1,3],[2,6],[8,10],[15,18]]
출력: [[1,6],[8,10],[15,18]]
설명: 구간 [1,3]과 [2,6]이 겹치므로 [1,6]으로 병합합니다.
예제 2:

입력: intervals = [[1,4],[4,5]]
출력: [[1,5]]
설명: 구간 [1,4]와 [4,5]는 겹치는 것으로 간주됩니다.

제약 조건:

1 <= intervals.length <= 104
intervals[i].length == 2
0 <= starti <= endi <= 104

```python
# 1. s, e 순서대로 sort, 이거 e 도 정렬해야할 거 같은데 어떻게 하는 건지 기억 안 남 ㅋㅎ 해결 
# 2. e가 다음의 s 보다 크면 머지 
# 새로운 e 가 되고 e가 다음의 s 보다 작으면 머지를 멈추고 저장
```

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        # 시작 시간, 종료 시간 순서대로 정렬
        intervals.sort(key=lambda x: (x[0], x[1]))
 
        # 초기값 세팅
        start_time = intervals[0][0]
        end_time = intervals[0][1]
        answer = []

        for s, e in intervals[1:]:
            if end_time >= s:
                # 머지 
                end_time = max(end_time, e)
            else: 
                # 머지 멈춤 
                answer.append([start_time, end_time])

                # 새로운 시작 시간, 종료 시간 세팅 필요
                # 다음 s, e 가 시작 시간, 종료 시간으로 세팅
                start_time = s
                end_time = e
        
        # 마지막 머지 저장
        answer.append([start_time, end_time])       

        return answer
            

```

```python
# sort 

intervals = [[1,3],[2,6], [1, 2], [8,10],[15,18]]

intervals.sort(key = lambda x: (x[0], x[1]))

print(intervals)
```

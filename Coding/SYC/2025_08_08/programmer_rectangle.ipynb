입출력 예 #3

rect_8.png

캐릭터 위치는 (1, 1)이며, 아이템 위치는 (4, 7)입니다. 위 그림과 같이 굵은 선을 따라 이동하는 경로가 가장 짧습니다.

```python
def solution(rectangle, characterX, characterY, itemX, itemY):
    # 102 x 102 크기 (스케일링 생략할 수 없지만, 외곽만 추적하는 방식)
    board = [[0] * 102 for _ in range(102)]
    
    for rec in rectangle:
        x1, y1, x2, y2 = [x * 2 for x in rec]
        for x in range(x1, x2 + 1):
            for y in range(y1, y2 + 1):
                board[y][x] = 1
    
    # 내부 0으로 초기화
    for rec in rectangle:
        x1, y1, x2, y2 = [x * 2 for x in rec]
        for x in range(x1 + 1, x2):
            for y in range(y1 + 1, y2):
                board[y][x] = 0

    # 외곽 경로 따라가는 방식: 방향 순서 유지
    dirs = [(1, 0), (0, 1), (-1, 0), (0, -1)]  # 우하좌상
    from collections import deque

    visited = [[False] * 102 for _ in range(102)]
    sx, sy = characterX * 2, characterY * 2
    ex, ey = itemX * 2, itemY * 2

    q = deque()
    q.append((sx, sy, 0))
    visited[sy][sx] = True

    while q:
        x, y, dist = q.popleft()

        if x == ex and y == ey:
            return dist // 2

        for dx, dy in dirs:
            nx, ny = x + dx, y + dy
            if 0 <= nx < 102 and 0 <= ny < 102:
                if board[ny][nx] == 1 and not visited[ny][nx]:
                    visited[ny][nx] = True
                    q.append((nx, ny, dist + 1))
```

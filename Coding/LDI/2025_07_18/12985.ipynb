# 예상 대진표

```python
# 첫번째 시도

def solution(n,a,b):
    answer = 1
    while abs(a-b) != 1:
        answer += 1
        if a % 2 == 0:
            a = a // 2
        else:
            a = a // 2 + 1
        if b % 2 == 0:
            b = b // 2
        else:
            b = b // 2 + 1
            
    
    return answer
```

* 제출해보면 몇 가지 테스트 케이스에서 틀렸다고 뜬다
* 이유를 생각해보니 a,b차이가 1인 경우 예를 들어, a=4, b=5이면 a와 b가 대전을 하기 위해서는 2번의 경기가 필요하지만 답은 1로 출력이 되는 오류가 발생

```python
# 정답

def solution(n,a,b):
    answer = 1
    lst = [a, b]
    lst = sorted(lst)
    while True:
        print(lst)
        if (lst[1] - lst[0]) == 1 and lst[1] % 2 == 0 :
            break
        answer += 1
        if lst[0] % 2 == 0:
            lst[0] = lst[0] // 2
        else:
            lst[0] = lst[0] // 2 + 1
        if lst[1] % 2 == 0:
            lst[1] = lst[1] // 2
        else:
            lst[1] = lst[1] // 2 + 1
        
    
    return answer
```

## 경주로 건설

```python
from collections import deque

def solution(board):
    N = len(board)
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]  
    
    cost_board = [[[float('inf')] * 4 for _ in range(N)] for _ in range(N)]
    
    queue = deque()
    
    for i in [1, 3]: 
        dx, dy = directions[i]
        nx, ny = dx, dy
        if 0 <= nx < N and 0 <= ny < N and board[nx][ny] == 0:
            cost_board[nx][ny][i] = 100
            queue.append((nx, ny, i, 100))
    
    while queue:
        x, y, direction, cost = queue.popleft()
        
        for i in range(4): 
            dx, dy = directions[i]
            nx, ny = x + dx, y + dy
            
            if not (0 <= nx < N and 0 <= ny < N):
                continue
            if board[nx][ny] == 1:
                continue
            
            new_cost = cost + (100 if i == direction else 600)
            
            if cost_board[nx][ny][i] > new_cost:
                cost_board[nx][ny][i] = new_cost
                queue.append((nx, ny, i, new_cost))
    
    return min(cost_board[N-1][N-1])
```

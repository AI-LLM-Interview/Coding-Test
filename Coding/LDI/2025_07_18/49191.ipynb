# 순위

* 이긴 선수 그래프 (win_graph) 와 진 선수 그래프 (lose_graph) 생성
* 각 선수마다 DFS를 통해 몇 명에게 이겼고, 졌는지 계산
* 이긴 수 + 진 수 = n - 1 이면 순위가 정해진 것

```python
def solution(n, results):
    from collections import defaultdict

    win_graph = defaultdict(list)
    lose_graph = defaultdict(list)

    for winner, loser in results:
        win_graph[winner].append(loser)
        lose_graph[loser].append(winner)

    def dfs(start, graph):
        visited = [False] * (n + 1)
        stack = [start]
        visited[start] = True
        count = 0

        while stack:
            current = stack.pop()
            for next_player in graph[current]:
                if not visited[next_player]:
                    visited[next_player] = True
                    stack.append(next_player)
                    count += 1
        return count

    answer = 0
    for i in range(1, n+1):
        win_count = dfs(i, win_graph)   # i가 이긴 사람 수
        lose_count = dfs(i, lose_graph) # i가 진 사람 수

        if win_count + lose_count == n - 1:
            answer += 1

    return answer
```

* 예시 : [[4, 3], [4, 2], [3, 2], [1, 2], [2, 5]]

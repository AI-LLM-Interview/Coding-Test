### 디스크 컨트롤러

```python
import heapq

def solution(jobs):
    jobs.sort()  # 요청시간 기준 정렬
    heap = []
    time = 0
    i = 0
    total = 0
    count = len(jobs)
    
    while i < count or heap:
        while i < count and jobs[i][0] <= time:
            heapq.heappush(heap, (jobs[i][1], jobs[i][0]))  # (소요시간, 요청시간)
            i += 1
        if heap:
            duration, start = heapq.heappop(heap)
            time += duration
            total += time - start
        else:
            time = jobs[i][0]
    
    return total // count
```

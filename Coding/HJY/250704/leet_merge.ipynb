```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        answer = []
        lines = []
        for interval in intervals:
            lines.append((interval[0], 0))
            lines.append((interval[1], 1))
        lines.sort()
        print(lines)
        start, end = -1, -1
        stack = []
        for v, se in lines:
            if se == 0:
                if stack == []:
                    start = v
                stack.append(1)
            else:
                stack.pop()
                if stack == []:
                    end = v
                    answer.append([start,end])
        return answer
```

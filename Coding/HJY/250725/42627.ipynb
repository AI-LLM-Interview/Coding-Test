```python
import heapq

def solution(jobs):
    size = len(jobs)
    answer = 0
    cur = 0
    queue = []
    jobs.sort()
    jobs = [(i, s, l) for i, (s,l) in enumerate(jobs)]
    while queue or jobs:
        while jobs and jobs[0][1] <= cur:
            i, s, l = jobs.pop(0)
            heapq.heappush(queue, (l, s, i))
        if not queue:
            cur = jobs[0][1]
        if queue:
            l, s, i = heapq.heappop(queue)
            cur += l
            answer += cur - s
    return answer//size
```

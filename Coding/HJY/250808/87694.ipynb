```python
def solution(rectangle, X, Y, itemX, itemY):
    answer = 0
    
    board = [[0]*101 for i in range(101)]
    for x1, y1, x2, y2 in rectangle:
        for i in range(x1*2, x2*2+1):
            for j in range(y1*2, y2*2+1):
                board[i][j] = 1
                
    for x1, y1, x2, y2 in rectangle:
        for i in range(x1*2+1, x2*2):
            for j in range(y1*2+1, y2*2):
                board[i][j] = 0

    def check(x,y):
        return 0 < x < 101 and 0 < y < 101 and board[x][y]

    queue = [(X*2, Y*2, 0)]
    visited = [(X*2,Y*2)]
    while queue:
        x,y,w = queue.pop(0)
        if x == itemX*2 and y == itemY*2:
            return w
        if (x+2,y) not in visited and check(x+1,y) and check(x+2,y):
            visited.append((x+2, y))
            queue.append((x+2, y, w+1))
        if (x-2,y) not in visited and check(x-1,y) and check(x-2,y):
            visited.append((x-2, y))
            queue.append((x-2, y, w+1))
        if (x,y+2) not in visited and check(x,y+1) and check(x,y+2):
            visited.append((x, y+2))
            queue.append((x, y+2, w+1))
        if (x,y-2) not in visited and check(x,y-1) and check(x,y-2):
            visited.append((x, y-2))
            queue.append((x, y-2, w+1))
    return answer
```

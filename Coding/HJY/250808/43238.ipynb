```python
def solution(n, times):
    left, right = 0, times[0] * n
    
    def calculate(t):
        return sum([t//time for time in times])
    
    time = calculate(right)
    
    while left <= right:
        t = (left + right) // 2
        result = calculate(t)
        if n <= result:
            time = t
            right = (left + right) // 2 - 1
        else:
            left = (left + right) // 2 + 1
    return time
```

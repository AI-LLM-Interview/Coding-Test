```python
from itertools import permutations

def solution(k, dungeons):
    answer = [-1]
    for dungeon in permutations(dungeons):
        prd = k
        count = 0
        for need, use in dungeon:
            if prd < need:
                break
            prd -= use
            count += 1
        answer.append(count)
    return max(answer)
```

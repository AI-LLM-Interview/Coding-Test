```python
from itertools import permutations

def solution(numbers):
    all_combinations = set()
    # 각 길이 1부터 len(numbers)까지의 조합 생성
    for i in range(1, len(numbers) + 1):
        all_combinations.update(int("".join(p)) for p in permutations(numbers, i))
    
    # 소수 판별 후, 소수의 개수를 반환
    return len(set(filter(is_prime, all_combinations)))

def is_prime(n):
    if n <= 1:
        return False
    s = 2
    while s * s <= n:
        if n % s == 0:
            return False
        s += 1
    return True
```

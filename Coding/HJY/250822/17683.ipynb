```python
def solution(m, musicinfos):
    answer = ""
    td = 0
    m = "".join(splitSound(m))
    for song in musicinfos:
        start, end, title, sound = song.split(",")
        arr = splitSound(sound)
        temp = ""
        time = timeDiff(start, end)
        for i in range(time):
            temp += arr[i%len(arr)]
        if m in temp:
            if td < time:
                td = time
                answer = title
    return answer or "(None)"

def timeDiff(start, end):
    h1, m1 = map(int, start.split(":"))
    h2, m2 = map(int, end.split(":"))
    return (h2 - h1) * 60 + m2 - m1

def splitSound(song):
    arr = []
    i = 0
    while i < len(song):
        if i+1 < len(song) and song[i+1] == "#":
            arr.append(song[i].lower())
            i += 2
        else:
            arr.append(song[i])
            i += 1
    return arr
```

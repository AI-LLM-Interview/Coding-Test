```python
def solution(h1, m1, s1, h2, m2, s2):
    answer = 0
    n = (h2 - h1)*3600 + (m2 - m1)*60 + s2 - s1
    hh = h1*5
    mm = m1
    ss = s1
    for _ in range(n):
        hh %= 60
        mm %= 60
        ss %= 60
        if hh <= ss <= hh + 1 and hh == mm:
            answer += 1
        elif hh <= ss <= hh + 1:
            answer += 1
        elif mm <= ss <= mm + 1:
            answer += 1
        hh += 1/3600
        mm += 1/60
        ss += 1
    return answer
```

```python
import heapq

def solution(n, works):
    answer = 0
    works = [-work for work in works]
    heapq.heapify(works)
    for _ in range(n): 
        a = heapq.heappop(works)
        if a == 0:
            return 0
        heapq.heappush(works, a+1)
    for work in works:
        answer += work**2
    return answer
```

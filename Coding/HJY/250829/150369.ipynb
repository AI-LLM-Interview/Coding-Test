```python
import math

def solution(cap, n, deliveries, pickups):
    answer = 0
    need_d = 0
    need_p = 0

    for i in range(n - 1, -1, -1):
        need_d += deliveries[i]
        need_p += pickups[i]
        # 이 지점까지 필요한 최소 왕복 수 (ceil 사용)
        trips = max(math.ceil(need_d / cap), math.ceil(need_p / cap))
        if trips > 0:
            answer += (i + 1) * 2 * trips
            need_d -= trips * cap
            need_p -= trips * cap
    return answer
```

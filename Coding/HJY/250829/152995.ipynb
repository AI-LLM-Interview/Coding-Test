```python
import heapq

def solution(scores):
    li = []
    wh = scores[0]
    for s1, s2 in scores[1:]:
        if wh[0] < s1 and wh[1] < s2:
            return -1
        if sum(wh) < s1 + s2:
            heapq.heappush(li, (-s1, s2))
    answer = len(li) + 1
    max_b = 0
    while li:
        a, b = heapq.heappop(li)
        if b < max_b:
            answer -= 1
        if max_b < b:
            max_b = b
            
    return answer
```

```python
def solution(n, computers):
    answer = 0
    visited = [False]*n
    for i in range(n):
        if not visited[i]:
            dfs(i, n, computers, visited)
            answer += 1
    return answer

def dfs(i, n, computers, visited):
    if not visited[i]:
        visited[i] = True
        for j in range(n):
            if computers[i][j] == 1 and j != i and not visited[j]:
                dfs(j, n, computers, visited)
```

```python
solution(3, [[1, 1, 0], [1, 1, 0], [0, 0, 1]])  # Example usage
```

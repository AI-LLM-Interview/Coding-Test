```python
from collections import deque

def solution(board):
    n = len(board)
    answer = float("inf")
    
    # (x, y, cost, direction, visited set)
    queue = deque([
        (0, 0, 0, "R", set([(0, 0)])),
        (0, 0, 0, "D", set([(0, 0)]))
    ])
    
    # 최소 비용 기록 (좌표 + 방향 기준)
    visited_cost = {}

    while queue:
        x, y, cost, direction, visited = queue.popleft()

        if (x, y) == (n - 1, n - 1):
            answer = min(answer, cost)
            continue

        # ➡ 오른쪽 이동
        if y + 1 < n and board[x][y + 1] == 0 and (x, y + 1) not in visited:
            new_cost = cost + (100 if direction == "R" else 600)
            key = (x, y + 1, "R")
            if key not in visited_cost or visited_cost[key] > new_cost:
                visited_cost[key] = new_cost
                new_visited = visited.copy()
                new_visited.add((x, y + 1))
                queue.append((x, y + 1, new_cost, "R", new_visited))

        # ⬅ 왼쪽 이동
        if y - 1 >= 0 and board[x][y - 1] == 0 and (x, y - 1) not in visited:
            new_cost = cost + (100 if direction == "L" else 600)
            key = (x, y - 1, "L")
            if key not in visited_cost or visited_cost[key] > new_cost:
                visited_cost[key] = new_cost
                new_visited = visited.copy()
                new_visited.add((x, y - 1))
                queue.append((x, y - 1, new_cost, "L", new_visited))

        # ⬇ 아래 이동
        if x + 1 < n and board[x + 1][y] == 0 and (x + 1, y) not in visited:
            new_cost = cost + (100 if direction == "D" else 600)
            key = (x + 1, y, "D")
            if key not in visited_cost or visited_cost[key] > new_cost:
                visited_cost[key] = new_cost
                new_visited = visited.copy()
                new_visited.add((x + 1, y))
                queue.append((x + 1, y, new_cost, "D", new_visited))

        # ⬆ 위 이동
        if x - 1 >= 0 and board[x - 1][y] == 0 and (x - 1, y) not in visited:
            new_cost = cost + (100 if direction == "U" else 600)
            key = (x - 1, y, "U")
            if key not in visited_cost or visited_cost[key] > new_cost:
                visited_cost[key] = new_cost
                new_visited = visited.copy()
                new_visited.add((x - 1, y))
                queue.append((x - 1, y, new_cost, "U", new_visited))

    return answer
```

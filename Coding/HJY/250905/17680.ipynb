```python
def solution(cacheSize, cities):
    if cacheSize == 0:
        return len(cities) * 5
    cities = list(map(str.lower, cities))
    cache = []
    answer = 0
    for city in cities:
        if city not in cache:
            if cacheSize == len(cache):
                cache.pop(0)
            answer += 5
            cache.append(city)
        else:
            answer += 1
            cache.remove(city)
            cache.append(city)
    return answer
```

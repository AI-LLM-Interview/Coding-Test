```python
def solution(citations):
    answer = 0
    for h in range(max(citations)+1):
        h_up = len(list(filter(lambda x: h<=x, citations)))
        h_down = len(list(filter(lambda x: x<=h, citations)))
        if h <= h_up and h_down <= h:
            answer = max(answer, h)
        elif answer != 0:
            break
    return answer
```

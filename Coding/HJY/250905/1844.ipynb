```python
from collections import deque

def solution(maps):
    n, m = len(maps), len(maps[0])
    answer = -1
    visited = set([(0,0)])
    queue = deque([(1,0,0)])
    while queue:
        w, x, y = queue.popleft()
        if (x, y) == (n-1, m-1):
            return w
        if x+1 < n and (x+1, y) not in visited and maps[x+1][y]:
            visited.add((x+1,y))
            queue.append((w+1, x+1, y))
        if y+1 < m and (x, y+1) not in visited and maps[x][y+1]:
            visited.add((x,y+1))
            queue.append((w+1, x, y+1))
        if 0 <= x-1 and (x-1, y) not in visited and maps[x-1][y]:
            visited.add((x-1,y))
            queue.append((w+1, x-1, y))
        if 0 <= y-1 and (x, y-1) not in visited and maps[x][y-1]:
            visited.add((x,y-1))
            queue.append((w+1, x, y-1))
    return answer
```

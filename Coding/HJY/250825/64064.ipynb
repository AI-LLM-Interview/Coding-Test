```python
import re

def solution(user_id, banned_id):
    answer = 0
    blacklist = {}
    for bid in banned_id:
        bid = bid.replace("*", ".")
        for uid in user_id:
            if len(bid) == len(uid) and re.match(bid, uid):
                if bid in blacklist.keys():
                    if uid not in blacklist[bid]:
                        blacklist[bid].append(uid)
                else:
                    blacklist[bid] = [uid]
    return answer
```

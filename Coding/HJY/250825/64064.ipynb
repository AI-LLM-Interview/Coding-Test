```python
import re

def solution(user_id, banned_id):
    size = len(banned_id)
    banned_id = [bid.replace("*", ".") for bid in banned_id]
    answers = []
    blacklist = {}
    for bid in banned_id:
        for uid in user_id:
            if len(bid) == len(uid) and re.match(bid, uid):
                if bid in blacklist.keys():
                    if uid not in blacklist[bid]:
                        blacklist[bid].append(uid)
                else:
                    blacklist[bid] = [uid]
    results = []
    for bid in banned_id:
        results.append((bid, blacklist[bid]))
    print(results)
    
    def func(idx, arr):
        if size == idx:
            # print(idx, arr, size)
            if len(arr) == size:
                answers.append(",".join(sorted(arr[:])))
            return
        for candidate in results[idx][1]:
            if candidate not in arr:
                arr.append(candidate)
                func(idx+1, arr)
                arr.pop()
        
    func(0, [])
    
    return len(set(answers))
```

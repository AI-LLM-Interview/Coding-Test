```python
def solution(numbers):
    answer = []
    for num in numbers:
        tree = bin(num)[2:]
        k = 0
        while 2**k - 1 < len(tree):
            k += 1
        tree = "0" * (2**k - 1 - len(tree)) + tree
        answer.append(int(isBinTree(tree)))
    return answer

def isBinTree(tree):
    mid = len(tree) // 2
    if tree.count("1") == 0:
        return True
    if tree[mid] == "0":
        return False
    if len(tree) <= 3:
        return True
    return isBinTree(tree[:mid]) and isBinTree(tree[mid+1:])
```

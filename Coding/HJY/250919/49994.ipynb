```python
def solution(dirs):
    answer = set()
    i, j = 0, 0
    for d in dirs:
        if d == "U" and -5 < i:
            answer.add((i-.5,j))
            i -= 1
        if d == "D" and i < 5:
            answer.add((i+.5,j))
            i += 1
        if d == "R" and j < 5:
            answer.add((i,j+.5))
            j += 1
        if d == "L" and -5 < j:
            answer.add((i,j-.5))
            j -= 1
    return len(answer)
```

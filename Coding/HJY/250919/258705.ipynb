```python
from itertools import combinations as comb
def solution(n, tops):
    total = 2*n + 1 + sum(tops)
    answer = 1
    case = []
    for i in range(n):
        case.append([2*i, 2*i+1])
        case.append([2*i+1, 2*i+2])
        if tops[i] == 1:
            case.append([2*i+1])
    for i in range(1, n+1):
        for c in comb(case, i):
            a = sum(c, [])
            b = len(set(a))
            a = len(a)
            if a == b:
                answer+=1
    return answer % 10007
```

```python
from itertools import combinations as comb
def solution(n, q, ans):
    candidates = list(comb([i+1 for i in range(n)], 5))
    answer = []
    for candidate in candidates:
        flag = True
        for row, a in zip(q, ans):
            cnt = 0
            for r in row:
                if r in candidate:
                    cnt += 1
            if cnt != a:
                flag = False
                break
        if flag:
            answer.append(candidate)
    return len(answer)
```

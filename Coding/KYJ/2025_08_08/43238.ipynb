```python
#  1, 2 빼고 다 실패
def solution(n, times):
    time = 0
    done = 0
    while done < n:
        time += 1
        for i in times:
            if time % i == 0:
                done += 1
                if done == n:
                    return time
    
    return time
```

```python
def solution(n, times):
    left = 1
    right = max(times) * n
    answer = right

    while left <= right:
        mid = (left + right) // 2
        total = sum(mid // t for t in times)  # mid 시간 동안 처리 가능한 사람 수
        print(total)
        if total >= n:  # 심사 가능하다면 시간 줄여보기
            answer = mid
            right = mid - 1
        else:  # 심사 부족 -> 시간 늘려야
            left = mid + 1
        print(left, mid, right)

    return answer
```

```python
from itertools import permutations as pt

def solution(numbers):
    answer = []
    numlist = list(numbers)
        
    for j in range(1,len(numbers)+1):#개수
        for k in pt(numlist, j):
            num = int(''.join(k))
            if num >= 2:
                chk = 0
                # for l in range(2,num//2 + 1):
                for l in range(2,int(num**0.5) + 1):
                    if num % l == 0:
                        chk = 1
                if chk == 0:
                    answer.append(num)
    return len(set(answer))
```

```python
from collections import deque

def solution(k, dungeons):
    n = len(dungeons)
    max_count = 0
    queue = deque()
    queue.append((k, [False] * n, 0)) 

    while queue:
        curr_k, visited, count = queue.popleft()
        max_count = max(max_count, count)

        for i in range(n):
            if not visited[i]:
                req, cost = dungeons[i]
                if curr_k >= req:
                    next_visited = visited[:]
                    next_visited[i] = True
                    queue.append((curr_k - cost, next_visited, count + 1))

    return max_count
```

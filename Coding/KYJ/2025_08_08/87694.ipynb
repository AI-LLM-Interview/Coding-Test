```python
k = 80
dungeons = [[80,20],[50,40],[30,10]]	
result = 3
```

```python
from collections import deque

def solution(k, dungeons):
    n = len(dungeons)
    max_count = 0
    queue = deque()
    queue.append((k, [False] * n, 0))   # 현재피로도, 방문 여부, 탐험 횟수

    while queue:
        curr_k, visited, count = queue.popleft()
        max_count = max(max_count, count) # 최대 방문 수

        for i in range(n):
            if not visited[i]:
                req, cost = dungeons[i]
                if curr_k >= req:
                    next_visited = visited[:]
                    next_visited[i] = True
                    queue.append((curr_k - cost, next_visited, count + 1))
        print(queue)

    return max_count
```

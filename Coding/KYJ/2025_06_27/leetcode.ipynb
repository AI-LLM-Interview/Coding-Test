42. Trapping Rain Water

Given n non-negative integers representing an elevation map where the width of each bar is 1, compute how much water it can trap after raining.

Example 1:

Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]

Output: 6

Explanation: The above elevation map (black section) is represented by array [0,1,0,2,1,0,1,3,2,1,2,1]. In this case, 6 units of rain water (blue section) are being trapped.

Example 2:

Input: height = [4,2,0,3,2,5]

Output: 9
 
Constraints:

n == height.length

1 <= n <= 2 * 104

0 <= height[i] <= 105

```python
height = [4,2,0,3,2,5]
```

```python
class Solution:
    def trap(self, height: List[int]) -> int:
        before_best = 0
        before_best_loc = 0
        water = 0
        for i in range(len(height)):
            if before_best == 0 and height[i] > before_best:
                before_best = height[i]
                before_best_loc = i
            elif before_best != 0 and height[i] >= before_best:
                water += (before_best * (i - before_best_loc)) - sum(height[before_best_loc:i])
                before_best = height[i]
                before_best_loc = i
            elif before_best != 0 and height[i] < before_best:
                # (1) 현재 위치 뒤쪽에서 가장 높은 기둥 높이
                rest_max = max(height[i + 1:], default=0)

                # (2) 오른쪽에 before_best 이상 기둥이 있으면 아직 계산하지 않고 진행
                if rest_max >= before_best:
                    continue

                # (3) 오른쪽에 before_best 이상 기둥이 없으면
                #     남은 구간의 최고 기둥(rest_max)을 새로운 left 벽으로 삼는다
                if rest_max == 0:          # 더 이상 벽이 없으면 물 못 채우므로 종료
                    break

                before_best = rest_max
                # rest_max는 반드시 존재하므로 .index() 안전
                before_best_loc = height.index(rest_max, i + 1)
                # 이후 for-loop가 자동으로 i를 1 증가시켜 계속 탐색
                continue
        return water
```

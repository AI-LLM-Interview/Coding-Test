```python
def solution(word):
    answer = 0
    word = word.replace("A","1").replace("E","2").replace("I","3").replace("O","4").replace("U","5")
    
    print(word)
    if len(word) == 5:
        answer += int(word[4])
    if len(word) >= 4:
        answer += (int(word[3])-1)*(5+1) + 1
    if len(word) >= 3:
        answer += (int(word[2])-1)*(25+5+1) + 1
    if len(word) >= 2:
        answer += (int(word[1])-1)*(125+25+5+1) + 1
    if len(word) >= 1:
        answer += (int(word[0])-1)*(625+125+25+5+1) +1
        
    return answer
```

```python
def solution(cap, n, deliveries, pickups):
    answer = 0
    
    while sum(deliveries) + sum(pickups) > 0:
        i = n - 1
        while i >= 0 and deliveries[i] == 0 and pickups[i] == 0:
            i -= 1
            
        far = i
        if far < 0:
            break

        answer += (far + 1) * 2

        deliver = cap
        pickup = cap

        i = far
        while i >= 0 and deliver > 0:
            if deliveries[i] == 0:
                i -= 1
                continue
            used = min(deliver, deliveries[i])   
            deliveries[i] -= used       
            deliver -= used                   
            if deliveries[i] == 0:
                i -= 1

        i = far
        while i >= 0 and pickup > 0:
            if pickups[i] == 0:
                i -= 1
                continue
            taken = min(pickup, pickups[i])      
            pickups[i] -= taken
            pickup -= taken
            if pickups[i] == 0:
                i -= 1

    return answer
```

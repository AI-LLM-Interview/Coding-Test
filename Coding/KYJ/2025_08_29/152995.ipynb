```python
def solution(scores):
    answer = 0
    a_list = []
    b_list = []
    sum_list = []
    for i in range(len(scores)):
        a_list.append(scores[i][0])
        b_list.append(scores[i][1])
        sum_list.append(scores[i][0] + scores[i][1])
        
    if a_list[0] == min(a_list) and b_list[0] == min(b_list):
        answer = -1
    else:
        rank = 0
        while answer == 0:
            now_max = max(sum_list)
            print(rank, now_max)
            if now_max != sum_list[0]:
                now_rank = sum_list.count(now_max)
                rank += now_rank
                for j in range(now_rank):
                    sum_list.remove(now_max)
            else:
                answer = rank+1
    return answer
```

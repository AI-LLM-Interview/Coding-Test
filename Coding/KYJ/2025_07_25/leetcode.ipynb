```python
class Solution:
    def convert(self, s: str, numRows: int) -> str:
        if len(s) == numRows or numRows == 1:
            return s
        else:
            for i in range(numRows):
                locals()[f'variable_{i+1}'] = []

            count = 1
            reverse = 0
            for j in range(len(s)):
                if reverse == 0:
                    locals()[f'variable_{count}'].append(s[j])
                    count += 1
                else:
                    locals()[f'variable_{count}'].append(s[j])
                    count -= 1

                if count == numRows and reverse == 0:
                    reverse = 1
                if count == 1 and reverse == 1:
                    reverse = 0
            answer = ""
            for k in range(numRows):
                answer += "".join(locals()[f'variable_{k+1}'])

            return answer
```

```python
lists = [[1,4,5],[1,3,4],[2,6]]
```

```python
len_list = []
for i in range(len(lists)):
    len_list.append(len(lists[i]))
max_len = max(len_list)
max_len
```

```python
answer = []
for j in range(max_len):
    for k in range(len(lists)):
        if len(lists[k]) >= j + 1:
            answer.append(lists[k][j])
```

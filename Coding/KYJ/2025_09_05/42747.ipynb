```python
def solution(citations):
    answer = 0
    n = len(citations)
    citations.sort()
    for i in range(0, max(citations) + 1):
        h = 0
        non = 0
        for j in range(n):
            if citations[j] >= i:
                h += 1
            else:
                non += citations[j]
        if h >= i:
            answer = max(answer, i) 
    return answer
```

```python
from collections import deque

def solution(board):
    answer = 0
    
    r_loc = []
    g_loc = []
    for i in range(len(board)):
        for j in range(len(board[0])):
            if board[i][j] == 'R':
                r_loc.append(i); r_loc.append(j)
            if board[i][j] == 'G':
                g_loc.append(i); g_loc.append(j)
    
    visit = [[999] * len(board[0]) for _ in range(len(board))]

    queue = deque([(r_loc[0], r_loc[1], 0)])
    visit[r_loc[0]][r_loc[1]] = 0
    
    while queue:
        now_x, now_y, count = queue.popleft()

        if count > visit[now_x][now_y]:
            continue

        for move in [(-1,0), (1,0), (0,-1), (0,1)]:
            move_x = now_x
            move_y = now_y
            while True:
                if not (0 <= move_x + move[0] < len(board) and 0 <= move_y + move[1] < len(board[0])) or board[move_x + move[0]][move_y + move[1]] == 'D':
                    break
                else:
                    move_x += move[0]
                    move_y += move[1]

            if move_x == now_x and move_y == now_y:
                continue

            if (count + 1) < visit[move_x][move_y]:
                visit[move_x][move_y] = count + 1
                queue.append((move_x, move_y, count + 1))

    if visit[g_loc[0]][g_loc[1]] != 999:
        return visit[g_loc[0]][g_loc[1]]
    else:
        return -1
```

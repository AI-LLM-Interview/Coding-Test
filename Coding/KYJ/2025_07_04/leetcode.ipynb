56. Merge Intervals

Given an array of intervals where intervals[i] = [starti, endi], merge all overlapping intervals, and return an array of the non-overlapping intervals that cover all the intervals in the input.

 

Example 1:

Input: intervals = [[1,3],[2,6],[8,10],[15,18]]

Output: [[1,6],[8,10],[15,18]]

Explanation: Since intervals [1,3] and [2,6] overlap, merge them into [1,6].

Example 2:

Input: intervals = [[1,4],[4,5]]

Output: [[1,5]]

Explanation: Intervals [1,4] and [4,5] are considered overlapping.
 

Constraints:

1 <= intervals.length <= 104

intervals[i].length == 2

0 <= starti <= endi <= 104

```python
intervals = [[1,3],[2,6],[8,10],[15,18]]
answer = []
intervals.sort()
tmp_list = []
tmp_list = intervals[0]
print(tmp_list)
```

```python
for i in range(1, len(intervals)):
    if tmp_list == [0, 0]:
        tmp_list = intervals[i]
    print(tmp_list)
    
    first_min = tmp_list[0]
    first_max = tmp_list[1]
    
    second_min = intervals[i][0]
    second_max = intervals[i][1]
    print(second_min, second_max)

    if (second_min >= first_min and second_min <= first_max) or (second_max >= first_min and second_max <= first_max):
        tmp_list = [min(first_min, second_min), max(first_max, second_max)]
    else:
        answer.append(tmp_list)
        tmp_list = [second_min, second_max]
    if i == len(intervals) - 1:
        answer.append(tmp_list)
```

코드 효율성 개선

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        answer = []
        intervals.sort()
        tmp_list = intervals[0]

        for i in range(1, len(intervals)):
            first_max = tmp_list[1]
            second_min, second_max = intervals[i]

            if second_min <= first_max:
                tmp_list[1] = max(first_max, second_max)
            else:
                answer.append(tmp_list)
                tmp_list = [second_min, second_max]
        answer.append(tmp_list)
```

```python
grid = [[1,2,3],[1,5,1],[4,2,1]]
grid = [[1,2,3],[4,5,6]]
```

```python
from collections import deque
# n = 8000000
n = len(grid)
m = len(grid[0])
dx = [1, 0]
dy = [0, 1]
INF = 8000000
cost = [[[INF]*2 for _ in range(m)] for _ in range(n)]
queue = deque()

for d in range(2):
    cost[0][0][d] = grid[0][0]
    queue.append((0, 0, 0, d))  

    while queue:
        x, y, c, d = queue.popleft()

        for i in range(2):
            nx = x + dx[i]
            ny = y + dy[i]
            if not (0 <= nx < n and 0 <= ny < m):
                continue
            print(nx, ny)
            new_cost = c + grid[nx][ny]

            if cost[nx][ny][i] > new_cost:
                cost[nx][ny][i] = new_cost
                queue.append((nx, ny, new_cost, i))
            print(queue)
```

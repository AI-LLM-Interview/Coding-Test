```python
def solution(h1, m1, s1, h2, m2, s2):
    def to_seconds(h, m, s):
        return h * 3600 + m * 60 + s

    t1 = to_seconds(h1, m1, s1)
    t2 = to_seconds(h2, m2, s2)
    if t2 < t1:
        t2 += 86400  # 자정 넘김

    sm = (59 * t2) // 3600 - (59 * t1) // 3600     
    sh = (719 * t2) // 43200 - (719 * t1) // 43200    

    both = (t2 // 43200) - (t1 // 43200) 
    ans = sm + sh - both

    start_overlap = ((59 * t1) % 3600 == 0) or ((719 * t1) % 43200 == 0)
    if start_overlap:
        ans += 1

    return ans
```

```python
def solution(user_id, banned_id):
    
    def match(u: str, b: str) -> bool:
        if len(u) != len(b):
            return False
        for ch_u, ch_b in zip(u, b):
            if ch_b != '*' and ch_u != ch_b:
                return False
        return True


    candidates = []
    for b in banned_id:
        cand = [i for i, u in enumerate(user_id) if match(u, b)]
        candidates.append(cand)


    result_masks = set()

    def dfs(idx: int, used_mask: int):
        if idx == len(candidates):
            result_masks.add(used_mask)
            return
        for i in candidates[idx]:
            if not (used_mask >> i) & 1:
                dfs(idx + 1, used_mask | (1 << i))


    candidates.sort(key=len)

    dfs(0, 0)
    return len(result_masks)
```

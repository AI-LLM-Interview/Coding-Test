```python
def solution(m, musicinfos):
    answer = ''
    def change(aa):
        return aa.replace("C#","c").replace("D#","d").replace("E#","e").replace("F#","f").replace("G#","g").replace("A#","a").replace("B#","b")
        
    m = change(m)
    time_chk = 0
    
    for i in range(len(musicinfos)):
        info = musicinfos[i].split(',')
        hour = int(info[1][:2]) - int(info[0][:2])
        minute = int(info[1][3:]) - int(info[0][3:])
        time = (hour * 60) + minute
        text = change(info[3])
        length = len(text)
        music = text * (time // length) + text[:time % length]
            
        if m in music and time > time_chk:
            answer = info[2]
            time_chk = time
    if time_chk == 0:
        answer = "(None)"
    return answer
```

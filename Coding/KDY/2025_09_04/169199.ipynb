리코쳇 로봇

```python
from collections import deque

def solution(board):
    # count = 0
    
    n,m = len(board), len(board[0])
    
    # 시작, 목표 위치 정의
    for i in range(n):
        for j in range(m):
            if board[i][j] == 'R':
                start = (i,j)
            if board[i][j] == 'G':
                goal = (i,j)
    
    visited = [[False]*m for _ in range(n)]
    # visited = [[0] * m for _ in range(n)]
    queue = deque()
    queue.append((start[0], start[1],0))
    visited[start[0]][start[1]] = True
    
    dx = [-1,1,0,0]
    dy = [0,0,-1,1]
    
    while queue:
        x,y, count = queue.popleft()
        
        if (x,y) == goal:
            return count
        
        for i in range(4):
            nx, ny = x, y
            
            # 벽에 부딪힐 때까지
            while True:
                
                tx = nx + dx[i]
                ty = ny + dy[i]
                
                if 0 <= tx < n and 0 <= ty < m and board[tx][ty] != 'D':
            # while (nx >= 0 or nx < n or ny >= 0 or ny < m) and board[nx][ny] != 'D':
                    nx = tx
                    ny = ty
                else:
                    break
                
            if not visited[nx][ny]:
                visited[nx][ny] = True
                queue.append((nx, ny, count + 1))

    return -1
```

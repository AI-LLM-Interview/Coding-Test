게임 맵 최단거리

```python
from collections import deque

def solution(maps):
    
    # answer = -1
    
    n,m = len(maps), len(maps[0])
    visited = [[0] * m for _ in range(n)]
    
    dx = [-1,1,0,0]
    dy = [0,0,-1,1]
    
    queue = deque()
    queue.append((0,0))
    visited[0][0] = 1
    
    while queue:
        x,y = queue.popleft()
        
        for i in range(4):
            nx = x + dx[i]
            ny = y + dy[i]
            
            if nx < 0 or nx >= n or ny < 0 or ny >= m:
                # 범위 밖이면 skip
                continue
                
            if maps[nx][ny] == 0 or visited[nx][ny]:
                # 벽이거나 이미 방문했다면 skip
                continue
                
            visited[nx][ny] = visited[x][y] + 1
            queue.append((nx,ny))   

    return visited[n-1][m-1] if visited[n-1][m-1] else -1
```

페이지 교체 기법
FIFO 선입선출
LRU 최근 참조
LFU 자주 참조

```python
def solution(cacheSize, cities):
    answer = 0
    cache = []
    
    if cacheSize == 0:
        return len(cities) * 5
    
    for name in cities:
        name= name.lower()
        if name in cache:
            cache.remove(name)
            cache.append(name)
            answer += 1
        else:
            if len(cache) >= cacheSize:
                cache.pop(0)
            cache.append(name)
            answer += 5
        
    return answer
```

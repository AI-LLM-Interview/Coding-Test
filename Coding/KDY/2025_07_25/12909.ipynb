# 올바른 괄호

```python
def solution(s):
    stack = []
    
    for txt in s:
        if txt == '(':
            stack.append('(')
        elif txt == ')':
            if stack:
                stack.pop()
            else:
                return False
    
    if stack:
        return False
    else:
        return True
```

```python
# test

solution("()()")
```

```python
def sol(n, grid, queries):
    a, b = 0, 0
    for i, query in enumerate(queries):
        x1, y1, x2, y2 = query
        for ii in range(x1, min(x1 + i + 1, n)):
            for jj in range(y1, min(y1 + i + 1, n)):
                a += grid[ii][jj]
                grid[ii][jj] = 0
        for ii in range(x2, min(x2 + i + 1, n)):
            for jj in range(y2, min(y2 + i + 1, n)):
                b += grid[ii][jj]
                grid[ii][jj] = 0
        
    return [a, b]
```

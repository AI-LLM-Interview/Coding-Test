# 올바른 괄호

```python
def solution(s):
    stack = []
    
    for txt in s:
        if txt == '(':
            stack.append('(')
        elif txt == ')':
            if stack:
                stack.pop()
            else:
                return False
    
    if stack:
        return False
    else:
        return True
```

```python
# test

solution("()()")
```

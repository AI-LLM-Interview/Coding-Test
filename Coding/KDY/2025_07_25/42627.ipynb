# 디스크 컨트롤러

```python
import heapq

def solution(jobs):
    answer = 0
    now = 0
    i = 0
    hq = []
    jobs.sort()
    n = len(jobs)
    
    print('1',jobs)
    
    while i < n or hq: # 수행한 작업을 꺼내 빈배열이면 모두 수행되는것
        # 현재 시점에서 작업할 수 있는 작업 모두 추가 (시점과 소요시간이 모두 같으면 먼저 들어간 작업번호가 수행되므로 작업번호는 고려x)
        
        # if job[0] >= now:
        #     heapq.heappush(hq, job)
        while i < n and jobs[i][0] <= now:
            # 요청된 작업리스트 = hq
            heapq.heappush(hq, [jobs[i][1], jobs[i][0]])
            i += 1
            
        # print('-',i, '\t hq',hq,'\t time',now)
            
        if not hq:
            now = jobs[i][0]
            
        if hq:
            # 현재 시점에서 작업 수행
            x=heapq.heappop(hq)
            # 작업 요청 시간 = x[1]
            # 작업 수행 시간 = x[0]
            now += x[0]

            # 반환 시각 = 작업 종료 - 요청시간 (now - x[0])
            answer += (now - x[1])
        
        
    return answer // n
        
```

```python
# Test cases
print(solution([[0, 3], [1, 9], [2, 6]]))  # Expected output: 9
print()
print(solution([[0, 5], [10, 3], [20, 2]]))  # Expected output: 3
```

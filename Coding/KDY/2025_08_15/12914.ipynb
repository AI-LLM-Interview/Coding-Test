멀리뛰기 V

```python
from itertools import permutations
from math import factorial

def solution(n):
    answer = 0
    for i in range((n//2)+1):
        n2 = i
        n1 = n-(2*n2)
        # print(n2)
        # print(n1)
        fn = factorial(n1+n2)
        fn1 = factorial(n1)
        fn2 = factorial(n2)
        answer += fn//(fn1*fn2)

    return answer% 1234567
```

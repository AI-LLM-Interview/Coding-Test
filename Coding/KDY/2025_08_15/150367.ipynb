```python
def tree_is(s):
    def check(sub):
        if len(sub) == 1:
            return True
        mid = len(sub) // 2
        root = sub[mid]
        left = sub[:mid]
        right = sub[mid+1:]
        if root == '0' and ('1' in left + right):
            return False
        return check(left) and check(right)
    return check(s)

def solution(numbers):
    answer = []
    for num in numbers:
        a = bin(num).replace('0b','')
        # print(a)
        length =len(str(a))
        k = 1
        while (2**k -1) < length:
            k += 1
        new_len = 2 **k -1
        s = '0' * (new_len - length) + str(a)
        # print(s)
        
        # 이진 트리인지 아닌지
        valid = 1 if tree_is(s) else 0
        answer.append(valid)
        
        
    return answer
```

```python
def tree_is(s, depth=0):
    def check(sub, depth):
        if len(sub) == 1:
            print("  " * depth, f"leaf: {sub}")
            return True
        mid = len(sub) // 2
        root = sub[mid]
        left = sub[:mid]
        right = sub[mid+1:]
        print("  " * depth, f"sub={sub}, root={root}, left={left}, right={right}")
        if root == '0' and ('1' in left + right):
            print("  " * depth, "❌ invalid: root=0 but child has 1")
            return False
        return check(left, depth+1) and check(right, depth+1)
    return check(s, depth)

def solution(numbers):
    answer = []
    for num in numbers:
        a = bin(num).replace('0b','')
        length = len(a)
        k = 1
        while (2**k - 1) < length:
            k += 1
        new_len = 2**k - 1
        s = '0' * (new_len - length) + a
        print(f"\nnum={num}, bin={a}, padded={s}")
        valid = 1 if tree_is(s) else 0
        print(f"✅ result for {num} -> {valid}")
        answer.append(valid)
    return answer

# 테스트
print(solution([7, 42, 5, 63, 95, 111 ]))
```

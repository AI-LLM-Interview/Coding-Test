야근지수

```python
import heapq

def solution(n, works):
    answer = 0
    lst = []
    for work in works:
        heapq.heappush(lst, -work)

    for _ in range(n):
        k = heapq.heappop(lst)
        if k == 0:
            return 0
        
        heapq.heappush(lst, k+1)
        # print(lst)
    
    for l in lst:
        answer += l**2

    return answer
```

피로도

|k	| dungeons	| result|
|--|--|--|
|80	| [[80,20],[50,40],[30,10]] |	3|

```python
from itertools import permutations


def solution(hp, dungeons):
    # answer = 0
    # [최소 필요, 소모 피로도]
    # hp
    # y=permutations(dungeons,len(dungeons))
    max_ = []
    
    # 모든 조합 중
    for dungeon in permutations(dungeons,len(dungeons)):
        # print(dungeon)
        answer = 0
        hp_ = hp
        
        # 하나의 조합에서 생기는 탐험 수
        for dun in dungeon:
            
            # print('dun1',dun[1], 'hp',hp_)
            
            if hp_ >= dun[0]:
                hp_ -= dun[1]
                answer += 1
            else:
                break
        max_.append(answer)
    return max(max_)
    # return max_
```

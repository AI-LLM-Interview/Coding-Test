입국심사

```python
def solution(n, times):
    answer = 0            
    left = 0
    right = min(times) * n
    
    while left <= right:
        t = (left + right) // 2
        ans = 0
        for time in times:
            ans += t // time
        if ans >= n: 
            answer = t
            right = t - 1
        else :
            # answer = t
            left = t + 1
    return answer
```

연습문제1

주어진 시간 t와 각 심사관의 처리 시간이 담긴 리스트 times가 있을 때,
모든 심사관이 t분 동안 처리할 수 있는 총 인원 수를 구하시오.

입력: t = 10, times = [7, 10]
출력: 2  (첫 심사관은 1명, 두 번째는 1명)

```python
t = 10
times = [7, 10]
answer = 0
for time in times: 
    ans = t // time 
    answer += ans

print(answer)  # 출력: 2
```

연습문제2

정렬된 리스트 arr와 값 target이 주어졌을 때,
target이 리스트에 있는지 이진탐색으로 확인해보세요.

입력: arr = [1, 3, 5, 7, 9], target = 5  
출력: True

```python
arr = [1, 2, 3, 4, 5, 9, 10]
target = 5
left = 0
right = len(arr) - 1
arr.sort()

while left <= right:
    mid = (left + right) // 2
    if arr[mid] == target:
        print(True)
        break
    elif arr[mid] > target:
        right = mid - 1
    else:
        left = mid + 1
else:
    print(False)

```

연습문제3

1부터 100 사이에서 어떤 수 x가 있을 때,
x 이상이면 조건이 참이 되는 상황에서
조건이 처음 참이 되는 최소값을 이진탐색으로 찾아보세요.

조건: x * x >= 30

출력: 6


```python
left = 1
right = 100
ans = 0
while left <= right:
    mid = (left + right) // 2
    x = mid **2
    if x >= 30:
        ans = mid
        right = mid -1
    else:
        left = mid + 1
        
print(ans)
```

소수찾기

예제 #2

[0, 1, 1]으로는 소수 [11, 101]를 만들 수 있습니다.

11과 011은 같은 숫자로 취급합니다.

```python
def solution(numbers):
    answer = 0
    # 모든 조합 만들고
    # 소수인지 확인

    numbers = [number for number in numbers]
    # print(numbers)

    # 조합
    from itertools import permutations

    digits = set()
    for i in range(1,len(numbers)+1):
        for p in permutations(numbers, i):
            # print(p)

            # set 중복 제거
            digits.add(int(''.join(p)))
    # print(digits)

    # 소수인지 확인
    for digit in digits:
        if digit in [0,1]:
            continue

        is_prime = True
        for k in range(2,int(digit**0.5)+1):
            if digit % k == 0:
                is_prime = False
                break

        if is_prime:   
            # print(digit)
            answer += 1
    return answer
```

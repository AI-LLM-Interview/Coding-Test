연습문제 1-2

직사각형의 좌측 하단 좌표가 (1,1), 우측 상단 좌표가 (3,3)일 때,  
이 직사각형의 테두리에 해당하는 좌표를 출력해보세요.

출력은 다음 순서를 따르세요:
→ 아래쪽 테두리 (왼→오)
→ 오른쪽 테두리 (아래→위)
→ 위쪽 테두리 (오→왼)
→ 왼쪽 테두리 (위→아래)

```python
x1, y1 = 1, 1
x2, y2 = 3, 3

border = []

# 아래쪽 테두리 (왼→오)
for x in range(x1, x2 + 1):
    border.append((x, y1))

# 오른쪽 테두리 (아래→위)
for y in range(y1 + 1, y2 + 1):
    border.append((x2, y))

# 위쪽 테두리 (오→왼)
for x in range(x2 - 1, x1 - 1, -1):
    border.append((x, y2))

# 왼쪽 테두리 (위→아래)
for y in range(y2 - 1, y1, -1):
    border.append((x1, y))

print(border)
```

연습문제 1

10x10 크기의 격자판이 있습니다.  
(좌측 하단이 (0,0)이고 우측 상단이 (9,9)입니다.)

정사각형 모양으로 테두리를 그려주세요.  
정사각형의 좌측 하단 좌표가 (2,2), 우측 상단 좌표가 (6,6)입니다.

이 테두리에 해당하는 좌표를 전부 리스트로 출력해보세요.

예시 출력 일부:
(2,2), (3,2), (4,2), ..., (6,6)


```python
dx = [1, 0, -1, 0]
dy = [0, 1, 0, -1]
# 순서: 오른쪽, 위쪽, 왼쪽, 아래쪽
x1, y1 = 2, 2  # 캐릭터의 초기 위치
x2, y2 = 6, 6  # 아이템의 위치
arr = []

def move(x, y, direction):
    return x + dx[direction], y + dy[direction]


for _ in range(4):
    arr.append((x1,y1))
    x1, y1 = move(x1, y1, 0)
    

for _ in range(4):
    arr.append((x1,y1))
    x1, y1 = move(x1, y1, 1)


for _ in range(4):
    arr.append((x1,y1))
    x1, y1 = move(x1, y1, 2)


for _ in range(4):
    arr.append((x1,y1))
    x1, y1 = move(x1, y1, 3)


print(arr)
```

연습문제 2

두 개의 직사각형이 다음 좌표를 가진다고 하자:

- 직사각형 A: [1,1,5,5]
- 직사각형 B: [3,3,7,7]

두 직사각형을 겹쳤을 때,  
**겉 테두리에만 해당하는 좌표들**을 리스트로 출력해보세요.

겹치는 안쪽 테두리는 포함하면 안 됩니다.


```python
# 두 직사각형의 테두리 좌표를 모두 구함
def get_border(x1, y1, x2, y2):
    border = set()
    for x in range(x1, x2 + 1):
        border.add((x, y1))
        border.add((x, y2))
    for y in range(y1 + 1, y2):
        border.add((x1, y))
        border.add((x2, y))
    return border

# 직사각형 A와 B의 테두리
border_a = get_border(1, 1, 5, 5)
border_b = get_border(3, 3, 7, 7)

# 전체 테두리 후보
all_border = border_a | border_b

# 두 직사각형의 내부 좌표(겹치는 부분 포함)를 구함
def get_inside(x1, y1, x2, y2):
    inside = set()
    for x in range(x1 + 1, x2):
        for y in range(y1 + 1, y2):
            inside.add((x, y))
    return inside

inside_a = get_inside(1, 1, 5, 5)
inside_b = get_inside(3, 3, 7, 7)
all_inside = inside_a | inside_b

# 테두리 중 내부에 포함되지 않는 것만 남김
outer_border = [pos for pos in all_border if pos not in all_inside]
# outer_border = sorted(outer_border)
print(outer_border)
```

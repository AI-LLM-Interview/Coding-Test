### 인사고과

```python
import heapq

def solution(scores):
    answer = 1
    x,y = scores[0]
    lst = []
    # print(x,y)
    
    # lst = scores[1:].sort(key=lambda t: (-t[0], t[1]))
    
    for m,n in scores[1:]:
        
        if m > x and n > y:
            return -1
        
        if x+y < m+n:
            answer += 1
            heapq.heappush(lst,[-m,n])
            
    b_max = -1
    while lst:
        a,b = heapq.heappop(lst)
        if b < b_max:
            answer -= 1
        elif b > b_max:
            b_max = b
        
    return answer
```

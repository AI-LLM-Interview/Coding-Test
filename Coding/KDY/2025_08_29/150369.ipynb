### 배달 수거하기

```python
# 시간초과

def solution(cap, n, deliveries, pickups):
    answer = 0
    remain_d = sum(deliveries)
    remain_p = sum(pickups)

    while remain_d > 0 or remain_p > 0:
        
        # 거리 계산
        far = 0
        for i in range(n-1,-1,-1):
            if deliveries[i] > 0 or pickups[i] > 0:
                far = i
                answer += (far+1) * 2
                break
        
        # 배달 계산
        truck_d = cap
        
        for j in range(far, -1, -1):
            if deliveries[j] > 0:
                t = min(deliveries[j], truck_d)
                deliveries[j] -= t
                truck_d -= t
                remain_d -= t
                # if deliveries[j] > truck_d:
                #     deliveries[j] -= truck_d
                #     truck_d = 0
                # else:
                #     truck_d -= deliveries[j]
                #     deliveries[j] = 0
            if truck_d == 0:
                break
                
                
        # 픽업 계산
        truck_p = cap
        
        for j in range(far, -1, -1):
            
            if pickups[j] > 0:
                tt = min(pickups[j], truck_p)
                pickups[j] -= tt
                truck_p -= tt
                remain_p -= tt
                
                # if pickups[j] > truck_p:
                #     pickups[j] -= truck_p
                #     truck_p = 0
                # else:
                #     truck_p -= pickups[j]
                #     pickups[j] = 0
            
            if truck_p == 0:
                break
                    
    return answer
```

```python
# 모범 답안

import math

def solution(cap, n, deliveries, pickups):
    answer = 0
    d_remain, p_remain = 0, 0  # 지금까지 처리해야 할 배달/수거 누적량
    
    # 가장 먼 집부터 앞으로 순회
    for i in range(n-1, -1, -1):
        d_remain += deliveries[i]
        p_remain += pickups[i]

        # 이번 집까지 남은 총량
        need = max(d_remain, p_remain)
        if need == 0:
            continue

        # 이번 집까지 몇 번 왕복해야 하나?
        trips = math.ceil(need / cap)
        answer += (i+1) * 2 * trips

        # 왕복으로 처리 가능한 만큼 줄여줌
        d_remain -= cap * trips
        p_remain -= cap * trips

    return answer
```
